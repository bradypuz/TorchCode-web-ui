# 🟡 Medium: REINFORCE Loss

Implement the **REINFORCE** policy-gradient loss for a batch of examples, each with one sampled action and one reward.

A policy is a model that assigns probabilities to possible actions. Each row of `logits` contains the policy's raw scores for those actions. `actions[i]` is the action already sampled from the current policy for example `i`; `rewards[i]` is the score received for taking it.

Compute the negative reward-weighted log probability of the chosen actions:

$$
\mathcal{L}_{\text{REINFORCE}} = -\frac{1}{B}\sum_{i=1}^{B} \operatorname{stop\_grad}(r_i)\,\log \pi_\theta(a_i \mid x_i).
$$

Positive rewards encourage the sampled actions; negative rewards discourage them. Zero rewards contribute no gradient. Treat rewards as fixed weights, so gradients flow only through `logits`.


Assume `B >= 1`, `A >= 1`, all tensors are on the same device, and action indices are valid. Use a numerically stable log-softmax over the action dimension, select one log probability per row, multiply by the detached rewards, then negate and average over the batch. Do not normalize rewards or subtract a baseline in this exercise.

This task implements the loss function, not a full sampling or training loop. In a multi-step reinforcement-learning problem, the weights would typically be returns associated with each sampled action.

# Solution: REINFORCE Loss

Reference solution using log-softmax and paired row/action indexing.

```python
import torch
import torch.nn.functional as F
from torch import Tensor
```

```python
# ✅ SOLUTION

def reinforce_loss(logits: Tensor, actions: Tensor, rewards: Tensor) -> Tensor:
    """REINFORCE loss with gradients through logits only.

    logits: (B, A) raw action scores
    actions: (B,) sampled action indices (torch.long)
    rewards: (B,) reward weights, treated as constants
    returns: scalar loss (Tensor)
    """
    logps = F.log_softmax(logits, dim=-1)
    rows = torch.arange(logps.shape[0], device=logps.device)
    chosen_logps = logps[rows, actions]
    return -(rewards.detach() * chosen_logps).mean()
```

```python
# Demo
logits = torch.tensor([[2., 1., 0.], [0., 1., 2.]], requires_grad=True)
actions = torch.tensor([0, 1])
rewards = torch.tensor([1., 0.])
print('Loss:', reinforce_loss(logits, actions, rewards).item())
# Expected: approximately 0.203803
```
